```python
import pandas as pd

data = {
    "hour": [1, 2, 3, 4, 5],
    "supply_air_temp_F": [55, 55.2, 55.1, 55.3, 62]
}

df = pd.DataFrame(data)
df
```

```python
df.plot(x="hour", y="supply_air_temp_F", marker="o")
```

```python
ax = df.plot(x="hour", y="supply_air_temp_F", marker="o")
ax.axhline(58, color="red", linestyle="--", label="Alarm limit")
ax.set_ylabel("Supply air temp (°F)")
ax.set_title("AHU-1 supply air temperature")
ax.legend()
```
